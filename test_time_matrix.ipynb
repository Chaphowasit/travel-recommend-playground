```python
import json
import os
import random
from dotenv import load_dotenv
import requests
import pandas as pd  # Import pandas for DataFrame support

load_dotenv()  # Load environment variables from .env file

# Load the activities data from a JSON file
with open(
    "./do/extract_json/job-do-1-300_output1_clean.json", "r", encoding="utf-8"
) as file:
    activities_json = json.load(file)

# Filter out locations with non-null latitude and longitude
valid_locations = [
    {
        "activity_name": activity["activity_name"],
        "latitude": activity["latitude"],
        "longitude": activity["longitude"],
    }
    for activity in activities_json
    if activity.get("latitude") is not None and activity.get("longitude") is not None
]

# Sample 5 random locations from the valid locations
sampled_locations = random.sample(valid_locations, k=min(5, len(valid_locations)))

# Prepare coordinates for the Mapbox Matrix API
coordinates = ";".join(
    f"{loc['longitude']},{loc['latitude']}" for loc in sampled_locations
)

# Prepare custom labels for the sampled locations
custom_labels = {i: loc["activity_name"] for i, loc in enumerate(sampled_locations)}

# Mapbox API details
access_token = os.getenv("MAPBOX_API_KEY")  # Fixed typo: changed os.env to os.getenv
url = "https://api.mapbox.com/directions-matrix/v1/mapbox/driving-traffic"
print(access_token)
# Prepare the request URL
request_url = f"{url}/{coordinates}?access_token={access_token}"

# Make the API request
response = requests.get(request_url)
data = response.json()

# # Output the results
# print("Sampled Locations:")
# for index, loc in custom_labels.items():
#     print(f"{index}: {loc}")

# print("\nCoordinates for API:")
# print(coordinates)

# print("\nAPI Response:")
# print(json.dumps(data, indent=4))

# Check if the response is OK
if data.get("code") == "Ok":
    print("\nTravel Time Matrix (in seconds):")
    durations = data["durations"]

    # Create a DataFrame from the durations
    df = pd.DataFrame(
        durations, columns=[custom_labels[i] for i in range(len(custom_labels))]
    )
    df.index = [
        custom_labels[i] for i in range(len(custom_labels))
    ]  # Set the index to the location names

    # Print the DataFrame neatly
    df.head(10)

else:
    print(f"Error: {data.get('message', 'Unknown error occurred.')}")
```
